```python
# Visualize regression vs classification difference
import matplotlib.pyplot as plt
import numpy as np

# Regression example
X = np.linspace(0, 10, 20)
y = 2 * X + 3 + np.random.randn(20)
plt.scatter(X, y)
plt.title("Regression: Predicting Continuous y")
plt.show()

# Classification example
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=100, n_features=2, n_redundant=0)
plt.scatter(X[:,0], X[:,1], c=y, cmap='bwr')
plt.title("Classification: Predicting Classes")
plt.show()
```

```python
import numpy as np, matplotlib.pyplot as plt
# Regression example
X = np.linspace(0, 10, 30)
y = 3*X + np.random.randn(30)*2
plt.subplot(1,2,1)
plt.scatter(X, y, c='blue'); plt.title("Regression (continuous)")

# Classification example
X = np.linspace(0,10,10)
y = (X > 5).astype(int)
plt.subplot(1,2,2)
plt.scatter(X, y, c=y, cmap='bwr'); plt.title("Classification (categories)")
plt.show()
```

```python
import numpy as np
# training data
x = np.array([2104, 1600, 2400, 1416, 3000])
y = np.array([400, 330, 369, 232, 540])

# Fit using Normal Equation (preview)
X = np.c_[np.ones(len(x)), x]  # add intercept term
theta = np.linalg.inv(X.T @ X) @ X.T @ y

print(f"θ₀ (intercept): {theta[0]:.2f}, θ₁ (slope): {theta[1]:.4f}")
# Predict
def predict(size): return theta[0] + theta[1]*size
print(predict(1250))
```
