## Sigmoid Function (Vectorized)

```python
def sigmoid(z):
    z = np.clip(z, -500,500)

    return 1 / (1 + np.exp(-z))
```

## Hypothesis Function (Vectorized)

```python
def hypothesis(X , w, b):
    z = X @ w + b
    
    return sigmoid(z)
```

## Cost Function(Log Loss)

```python
def compute_cost(h, y):

    m = len(y)
    eps = 1e-15

    h = np.clip(h,eps, 1-eps)

    cost = - (1/m) * np.sum(y*np.log(h) + (1-y)*np.log(1-h))
    return cost
```

## Gradients

```python
def compute_gradients(X, h, y):

    m = len(y)
    error = h - y 

    dw = (1 / m) * (X.T @ error)
    db = (1/m) * np.sum(error)  

    return dw, db
```

## Gradient Descent Loop

```python
def gradient_descent(X, y, w, b, learning_rate, iterations):
    m = len(y)
    
    cost_history = []
    w_history = []

    for i in range(iterations):

        h = hypothesis(X, w, b)

        cost = compute_cost(h, y)


        dw, db = compute_gradients(X, h, y)

        w = w - learning_rate * dw
        b = b - learning_rate * db

        cost_history.append(cost)
        w_history.append(w.copy())

        if i % 100 == 0:
            print(f"Iteration {i}, Cost: {cost}")

    return w, b, cost_history, w_history
```

## Predict Probabiltiy and Predict Class

```python
def predict_proba(X, w,b):
    return hypothesis(X, w, b)

def predict_class(X, w, b, threshold = 0.5):

    probs = predict_proba(X, w, b)
    return (probs >= threshold).astype(int)
```

## Accuracy Calculation

```python
def accuracy_score_manual(y_true, y_pred):
    return np.mean(y_true == y_pred)
```

```python
def confusion_matrix_manual(y_true, y_pred):
    tp = np.sum((y_pred == 1) & (y_true == 1))
    tn = np.sum((y_pred == 0) & (y_true == 0))
    fp = np.sum((y_pred == 1) & (y_true == 0))
    fn = np.sum((y_pred == 0) & (y_true == 1))
    
    return tp, tn, fp, fn
```

```python
def precision_manual(tp, fp):
    eps = 1e-15
    return tp / (tp + fp + eps)
```

```python
precision = precision_manual(tp, fp)
print("Precision:", precision)
```
